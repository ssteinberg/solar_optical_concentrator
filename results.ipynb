```python
# hello world
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rc('figure', figsize = (6, 6))
```

```python
# import data
data_coords = np.loadtxt('build/data_coords.csv', delimiter = ',')
data_rays_pos = np.loadtxt('build/data_rays_pos.csv', delimiter = ',')
data_rays_neg = np.loadtxt('build/data_rays_neg.csv', delimiter = ',')
```

```python
# plot circle at origin
circle = plt.Circle((0, 0), 0.0045, color = 'black')
plt.gca().add_patch(circle)

# plot mirrors
plt.plot(data_coords[:, 0], data_coords[:, 1], c = 'black')
plt.plot(data_coords[:, 2], data_coords[:, 3], c = 'black')

# plot positive extreme ray paths
# for row in data_rays_pos:
#     X = [x for j, x in enumerate(row) if j % 2 == 0]
#     Y = [y for j, y in enumerate(row) if j % 2 == 1]
#     plt.plot(X, Y, 'r', linewidth = 0.25)
for row in data_rays_pos:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'r', linewidth = 0.25)

# plot negative extreme ray paths
# for row in data_rays_neg:
#     X = [x for j, x in enumerate(row) if j % 2 == 0]
#     Y = [y for j, y in enumerate(row) if j % 2 == 1]
#     plt.plot(X, Y, 'b', linewidth = 0.25)
for row in data_rays_neg:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'b', linewidth = 0.25)

# display
plt.axis('square')
plt.axis([-0.01, 0.01, -0.01, 0.01])
plt.show()
```
