```python
# hello world!
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rc('figure', figsize = (6, 6))
```

```python
# import data
data_coords = np.loadtxt('data_coords.csv', delimiter = ',')
data_rays = np.loadtxt('data_rays.csv', delimiter = ',')
```

```python
# plot circle at origin
circle = plt.Circle((0, 0), 0.0045, color = 'black')
plt.gca().add_patch(circle)

# sort by second column and plot mirrors
sorted = data_coords[data_coords[:, 1].argsort()]
plt.plot(sorted[:, 0], sorted[:, 1])
plt.plot(sorted[:, 4], sorted[:, 5])

# plot full ray paths
# for i in range(0, data_rays.shape[0], 2):
#     X_pos = [x for j, x in enumerate(data_rays[i]) if j % 2 == 0]
#     Y_pos = [y for j, y in enumerate(data_rays[i]) if j % 2 == 1]
#     plt.plot(X_pos, Y_pos, 'r', linewidth = 0.25)
#     X_neg = [x for j, x in enumerate(data_rays[i + 1]) if j % 2 == 0]
#     Y_neg = [y for j, y in enumerate(data_rays[i + 1]) if j % 2 == 1]
#     plt.plot(X_neg, Y_neg, 'b', linewidth = 0.25)
    
# plot final ray segment through target
for i in range(0, data_rays.shape[0], 2):
    X_pos = [data_rays[i, 4], data_rays[i, 6]]
    Y_pos = [data_rays[i, 5], data_rays[i, 7]]
    plt.plot(X_pos, Y_pos, 'r', linewidth = 0.5)
    X_neg = [data_rays[i + 1, 4], data_rays[i + 1, 6]]
    Y_neg = [data_rays[i + 1, 5], data_rays[i + 1, 7]]
    plt.plot(X_neg, Y_neg, 'b', linewidth = 0.5)

# display
plt.axis('square')
plt.axis([-0.01, 0.01, -0.01, 0.01])
plt.show()
```
