```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
EPSILON, inv, L, f, dB, B_max = np.loadtxt('datainfarb/input.csv', delimiter = ',')
mirror1a = np.loadtxt('datainfarb/mirror1a.csv', delimiter = ',')
mirror1b = np.loadtxt('datainfarb/mirror1b.csv', delimiter = ',')
mirror2a = np.loadtxt('datainfarb/mirror2a.csv', delimiter = ',')
mirror2b = np.loadtxt('datainfarb/mirror2b.csv', delimiter = ',')
inverted_text = "inverted" if inv else "not inverted"
print(f"infinite system, arbitrary target, {inverted_text}, L = {L}, f = {f}, dB = {dB :.6f}, B_max = {B_max}")
print(f"epsilon = {EPSILON}")
```

```python
extr1 = np.loadtxt('datainfarb/extrinf1.csv', delimiter = ',')
extr2 = np.loadtxt('datainfarb/extrinf2.csv', delimiter = ',')

fig, ax = plt.subplots(1, figsize = (10, 10))

ax.plot([0, 0], [-0.125, 0.125], linewidth = 0.2)

ax.plot(mirror1a[:, 0], mirror1a[:, 1], linewidth = 2, c = 'blue', label = 'mirror 1a')
ax.plot(mirror1b[:, 0], mirror1b[:, 1], linewidth = 2, c = 'cornflowerblue', label = 'mirror 1b')
ax.plot(mirror2a[:, 0], mirror2a[:, 1], linewidth = 2, c = 'purple', label = 'mirror 2a')
ax.plot(mirror2b[:, 0], mirror2b[:, 1], linewidth = 2, c = 'mediumpurple', label = 'mirror 2b')

for row in extr1: ax.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'blue')
for row in extr2: ax.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'red')

height = -1
size = 0.01
# plt.axis([-size, size, -size + height, size + height])
zoom = 0.15
# plt.axis([-zoom, zoom, -zoom, zoom])
plt.show()
```
