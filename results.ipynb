```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
```

# Two-Mirror Concentrator

```python
data_2mc = np.loadtxt('build/data_2mc.csv', delimiter = ',')
data_mean_paths = np.loadtxt('build/data_mean_paths.csv', delimiter = ',')
data_pos_extr_paths = np.loadtxt('build/data_pos_extr_paths.csv', delimiter = ',')
data_neg_extr_paths = np.loadtxt('build/data_neg_extr_paths.csv', delimiter = ',')
```

```python
plt.scatter(0, 0, c = 'r', label = 'target')
plt.plot(data_2mc[:, 0], data_2mc[:, 1], linewidth = 2, c = 'purple', label = 'M1')
plt.plot(data_2mc[:, 2], data_2mc[:, 3], linewidth = 2, c = 'blue', label = 'M2')
plt.axis('square')
plt.legend()
plt.show()
```

# Concentration of mean rays

```python
plt.axis('square')
zoom = 0.00001
plt.axis([-zoom, zoom, -zoom, zoom])

circle = plt.Circle((0, 0), zoom, color = 'black')
plt.gca().add_patch(circle)

for row in data_mean_paths:
    # X = [x for j, x in enumerate(row) if j % 2 == 0]
	# Y = [y for j, y in enumerate(row) if j % 2 == 1]
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'g', linewidth = 0.25)

plt.show()
```

# Concentration of extreme rays

```python
plt.axis('square')
zoom = 0.01
plt.axis([-zoom, zoom, -zoom, zoom])

circle = plt.Circle((0, 0), zoom / 2.1, color = 'black')
plt.gca().add_patch(circle)

for row in data_pos_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'r', linewidth = 0.25)

for row in data_neg_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'b', linewidth = 0.25)

plt.show()
```

# Parabolic Mirror

```python
data_pm = np.loadtxt('build/data_pm.csv', delimiter = ',')
data_pm_a = np.loadtxt('build/data_pm_a.csv', delimiter = ',')
data_pm_b = np.loadtxt('build/data_pm_b.csv', delimiter = ',')
data_pm_c = np.loadtxt('build/data_pm_c.csv', delimiter = ',')
```

```python
plt.scatter(0, 0, c = 'black')
plt.plot(data_pm[:, 0], data_pm[:, 1], linewidth = 2, c = 'm')

for row in data_pm_a: plt.plot([row[2], row[4]], [row[3], row[5]], 'g', linewidth = 0.25)
for row in data_pm_b: plt.plot([row[2], row[4]], [row[3], row[5]], 'r', linewidth = 0.25)
for row in data_pm_c: plt.plot([row[2], row[4]], [row[3], row[5]], 'b', linewidth = 0.25)

plt.axis('square')
zoom = 0.005
plt.axis([-zoom, zoom, -zoom, zoom])
plt.show()
```
