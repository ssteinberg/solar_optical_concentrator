```python
import numpy as np
import matplotlib.pyplot as plt
```

# Two-Mirror Concentrator

```python
# import
# data_2mc = np.loadtxt('data_2mc_flat.csv', delimiter = ',')
# data_2mc = np.loadtxt('data_2mc_cyl.csv', delimiter = ',')
# data_2mc = np.loadtxt('data_scale_inv.csv', delimiter = ',')
data_2mc = np.loadtxt('data_2mc_fin_cyl.csv', delimiter = ',')

# source and target
plt.scatter(-10, 0, c = 'orange', label = 'source')
plt.scatter(0, 0, c = 'red', label = 'target')

# parabola
# Y = np.linspace(0, 2, 100)
# X1 = 0.256 * Y ** 2 - 2
# plt.plot(X1, Y, linewidth = 5, c = 'black')
# X2 = - 0.256 * Y ** 2 + 1
# plt.plot(X2, Y, linewidth = 5, c = 'black')



# fitting a polynomial
a, b, c = np.polyfit(data_2mc[:, 1], data_2mc[:, 0], deg = 2)
parabola = np.poly1d([a, b, c])
y_fit = np.linspace(data_2mc[0, 1], data_2mc[-1, 1], 100)
x_fit = parabola(y_fit)
plt.plot(x_fit, y_fit, linewidth = 5, c = 'black')



# M1
plt.plot(data_2mc[:, 0], data_2mc[:, 1], linewidth = 2, c = 'purple', label = 'M1')
# plt.text(data_2mc[0, 0], data_2mc[0, 1], s = f'({data_2mc[0, 0]}, {data_2mc[0, 1]})')
# plt.text(data_2mc[data_2mc.shape[0] - 1, 0], data_2mc[data_2mc.shape[0] - 1, 1], s = f'({data_2mc[data_2mc.shape[0] - 1, 0]}, {data_2mc[data_2mc.shape[0] - 1, 1]})')

# M2
plt.plot(data_2mc[:, 2], data_2mc[:, 3], linewidth = 2, c = 'blue', label = 'M2')
# plt.text(data_2mc[0, 2], data_2mc[0, 3], s = f'({data_2mc[0, 2]}, {data_2mc[0, 3]})')
# plt.text(data_2mc[data_2mc.shape[0] - 1, 2], data_2mc[data_2mc.shape[0] - 1, 3], s = f'({data_2mc[data_2mc.shape[0] - 1, 2]}, {data_2mc[data_2mc.shape[0] - 1, 3]})')

# display
plt.axis('equal')
plt.legend()
plt.show()
```

# Concentration of mean rays

```python
# import
# data_mean_paths = np.loadtxt('data_2mc_flat_mean.csv', delimiter = ',')
data_mean_paths = np.loadtxt('data_2mc_cyl_mean.csv', delimiter = ',')

# zoom
plt.axis('square')
zoom = 0.001
plt.axis([-zoom, zoom, -zoom, zoom])

# circle
circle = plt.Circle((0, 0), zoom, color = 'black')
plt.gca().add_patch(circle)

# mean ray paths
for row in data_mean_paths:
    # X = [x for j, x in enumerate(row) if j % 2 == 0]
	# Y = [y for j, y in enumerate(row) if j % 2 == 1]
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'g', linewidth = 0.25)

# display
plt.show()
```

# Concentration of extreme rays

```python
# import
# data_pos_extr_paths = np.loadtxt('data_2mc_flat_extr_pos.csv', delimiter = ',')
# data_neg_extr_paths = np.loadtxt('data_2mc_flat_extr_neg.csv', delimiter = ',')
data_pos_extr_paths = np.loadtxt('data_2mc_cyl_extr_pos.csv', delimiter = ',')
data_neg_extr_paths = np.loadtxt('data_2mc_cyl_extr_neg.csv', delimiter = ',')

# zoom
plt.axis('square')
zoom = 0.01
plt.axis([-zoom, zoom, -zoom, zoom])

# circle
circle = plt.Circle((0, 0), zoom / 2.1, color = 'black')
plt.gca().add_patch(circle)

# extreme ray paths
for row in data_pos_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'r', linewidth = 0.25)
for row in data_neg_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'b', linewidth = 0.25)

# display
plt.show()
```
