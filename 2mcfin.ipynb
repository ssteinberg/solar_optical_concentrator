```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('figure', figsize = (10, 6))
```

# Two-Mirror Concentrator

```python
# import
flat_source, flat_target, r, f1, L, f2, da, a_max = np.loadtxt('data_2mc_fin_cyl_input.csv', delimiter = ',')
data_2mc_m1 = np.loadtxt('data_2mc_fin_cyl_m1.csv', delimiter = ',')
data_2mc_m2 = np.loadtxt('data_2mc_fin_cyl_m2.csv', delimiter = ',')
data_paths = np.loadtxt('data_2mc_fin_cyl_paths.csv', delimiter = ',')

# source and target
if flat_source: plt.plot([-L, -L], [-r, r], linewidth = 5, color = 'orange', label = 'source')
else: plt.gca().add_patch(plt.Circle((-L, 0), r, color = 'orange', label = 'source'))
if flat_target: plt.plot([0, 0], [-r, r], linewidth = 5, color = 'red', label = 'target')
else: plt.gca().add_patch(plt.Circle((0, 0), r, color = 'red', label = 'target'))

# sort by second column and plot mirrors
data_2mc_m1_sorted = data_2mc_m1[data_2mc_m1[:, 1].argsort()]
plt.plot(data_2mc_m1_sorted[:, 0], data_2mc_m1_sorted[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1')
data_2mc_m2_sorted = data_2mc_m2[data_2mc_m2[:, 1].argsort()]
plt.plot(data_2mc_m2_sorted[:, 0], data_2mc_m2_sorted[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2')

# parabolas
M1Y = np.linspace(data_2mc_m1_sorted[0, 1], data_2mc_m1_sorted[-1, 1], 50)
M1X = 1 / (4 * f1) * (M1Y ** 2) - L - f1
plt.scatter(M1X, M1Y, s = 5, c = 'blue', label = 'parabola 1')
M2Y = np.linspace(data_2mc_m2_sorted[0, 1], data_2mc_m2_sorted[-1, 1], 50)
M2X = -1 / (4 * f2) * (M2Y ** 2) + f2
plt.scatter(M2X, M2Y, s = 5, c = 'purple', label = 'parabola 2')

# paths
# for row in data_paths: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'green')

# display
plt.axis('equal')
plt.legend()
plt.show()
```

# Extreme Rays

```python
data_extr_1 = np.loadtxt('data_2mc_fin_extr_1.csv', delimiter = ',')
data_extr_2 = np.loadtxt('data_2mc_fin_extr_2.csv', delimiter = ',')

if flat_source: plt.plot([-L, -L], [-r, r], linewidth = 5, color = 'orange', label = 'source')
else: plt.gca().add_patch(plt.Circle((-L, 0), 0.97 * r, color = 'orange', label = 'source'))
if flat_target: plt.plot([0, 0], [-r, r], linewidth = 5, color = 'orange', label = 'target')
else: plt.gca().add_patch(plt.Circle((0, 0), 0.97 * r, color = 'orange', label = 'target'))

for row in data_extr_1: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'red')
for row in data_extr_2: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'blue')

plt.axis('square')
scale = 1.25
# plt.axis([-L - scale * r, -L + scale * r, 0 - scale * r, 0 + scale * r])
plt.axis([-scale * r, scale * r, -scale * r, scale * r])
plt.show()
```
