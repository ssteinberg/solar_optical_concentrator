# Two-Mirror Concentrator

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rc('figure', figsize = (6, 6))
```

```python
data_coords = np.loadtxt('build/data_coords.csv', delimiter = ',')
data_mean_paths = np.loadtxt('build/data_mean_paths.csv', delimiter = ',')
data_pos_extr_paths = np.loadtxt('build/data_pos_extr_paths.csv', delimiter = ',')
data_neg_extr_paths = np.loadtxt('build/data_neg_extr_paths.csv', delimiter = ',')
```

```python
plt.plot(0, 0, marker = '.', c = 'r')
plt.plot(data_coords[:, 0], data_coords[:, 1], c = 'black')
plt.plot(data_coords[:, 2], data_coords[:, 3], c = 'black')
plt.axis('square')
plt.show()
```

# Concentration of mean rays

```python
plt.axis('square')
zoom = 0.00001
plt.axis([-zoom, zoom, -zoom, zoom])

circle = plt.Circle((0, 0), zoom, color = 'black')
plt.gca().add_patch(circle)

for row in data_mean_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'g', linewidth = 0.25)

plt.show()
```

# Concentration of extreme rays

```python
plt.axis('square')
zoom = 0.01
plt.axis([-zoom, zoom, -zoom, zoom])

circle = plt.Circle((0, 0), zoom / 2.1, color = 'black')
plt.gca().add_patch(circle)

for row in data_pos_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'r', linewidth = 0.25)

for row in data_neg_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'b', linewidth = 0.25)

plt.show()
```

# Notes:
- decreasing step size improves ray tracing and concentration
- increasing the scale of the system shifts the concentration along the x direction
