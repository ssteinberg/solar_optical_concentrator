```python
# hello world!
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rc('figure', figsize = (10, 10))
```

```python
# import data
data_coords = np.loadtxt('data_coords.csv', delimiter = ',')
data_rays = np.loadtxt('data_rays.csv', delimiter = ',')
```

```python
# plot origin
# plt.plot(0, 0, marker = '.', c = 'black', label = 'origin')

# plot circle
circle = plt.Circle((0, 0), 0.005, color = 'black')
plt.gca().add_patch(circle)

# plot coordinates
for row in data_coords:
    M1_X = [row[0], row[2]]
    M1_Y = [row[1], row[3]]
    plt.plot(M1_X, M1_Y, c = 'black', linewidth = 1)
    M2_X = [row[4], row[6]]
    M2_Y = [row[5], row[7]]
    plt.plot(M2_X, M2_Y, c = 'black', linewidth = 1)

# plot ray paths
for i in range(0, data_rays.shape[0], 2):
    X_pos = [x for j, x in enumerate(data_rays[i]) if j % 2 == 0]
    Y_pos = [y for j, y in enumerate(data_rays[i]) if j % 2 == 1]
    plt.plot(X_pos, Y_pos, 'r', linewidth = 0.25)
    X_neg = [x for j, x in enumerate(data_rays[i + 1]) if j % 2 == 0]
    Y_neg = [y for j, y in enumerate(data_rays[i + 1]) if j % 2 == 1]
    plt.plot(X_neg, Y_neg, 'b', linewidth = 0.25)

# display
plt.axis('square')
plt.axis([-0.01, 0.01, -0.01, 0.01]) # zoom in
plt.show()

# takes a couple of seconds because i am plotting all line segments ...
```

```python
# check Fermat's principle
for row in data_rays:
    p1 = np.array([row[0], row[1]])
    p2 = np.array([row[2], row[3]])
    p3 = np.array([row[4], row[5]])
    p4 = np.array([row[6], row[7]])
    print(np.linalg.norm(p2 - p1) + np.linalg.norm(p3 - p2) + np.linalg.norm(np.array([0, 0]) - p3))
```
