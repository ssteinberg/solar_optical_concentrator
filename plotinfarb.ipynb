```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('figure', figsize = (10, 5))
```

```python
inv, L, f, dB, B_max = np.loadtxt('datainfarb/input.csv', delimiter = ',')
mirror1a = np.loadtxt('datainfarb/mirror1a.csv', delimiter = ',')
mirror1b = np.loadtxt('datainfarb/mirror1b.csv', delimiter = ',')
mirror2a = np.loadtxt('datainfarb/mirror2a.csv', delimiter = ',')
mirror2b = np.loadtxt('datainfarb/mirror2b.csv', delimiter = ',')

inverted_text = "inverted" if inv else "not inverted"
print(f"infinite system, arbitrary target, {inverted_text}, L = {L}, f = {f}, dB = {dB :.6f}, B_max = {B_max}")
```

```python
plt.plot([0, 0], [-0.005, 0.005], c = 'orange')
# plt.plot([1, -2, 0, 1], [1, 0, -1, 1])

plt.plot(mirror1a[:, 0], mirror1a[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1a')
plt.plot(mirror1b[:, 0], mirror1b[:, 1], linewidth = 1, c = 'cornflowerblue', label = 'mirror 1b')
plt.plot(mirror2a[:, 0], mirror2a[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2a')
plt.plot(mirror2b[:, 0], mirror2b[:, 1], linewidth = 1, c = 'mediumpurple', label = 'mirror 2b')

plt.axis('equal')
plt.show()
```

```python
extr1 = np.loadtxt('datainfarb/extrinf1.csv', delimiter = ',')
extr2 = np.loadtxt('datainfarb/extrinf2.csv', delimiter = ',')

plt.plot([0, 0], [-0.005, 0.005], c = 'orange')

plt.plot(mirror1a[:, 0], mirror1a[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1a')
plt.plot(mirror1b[:, 0], mirror1b[:, 1], linewidth = 1, c = 'cornflowerblue', label = 'mirror 1b')
plt.plot(mirror2a[:, 0], mirror2a[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2a')
plt.plot(mirror2b[:, 0], mirror2b[:, 1], linewidth = 1, c = 'mediumpurple', label = 'mirror 2b')

for row in extr1: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'blue')
for row in extr2: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'red')

# plt.axis('equal')
plt.axis('square')
# zoom = 0.01
# plt.axis([-zoom, zoom, -zoom, zoom])
zoom = 0.005
plt.axis([-zoom, zoom, 0.005 - zoom, 0.005 + zoom])
plt.show()
```
