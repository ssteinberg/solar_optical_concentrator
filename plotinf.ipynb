```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('figure', figsize = (10, 5))
```

```python
flat, inv, L, f, dB, B_max = np.loadtxt('datainf/input.csv', delimiter = ',')
mirror1a = np.loadtxt('datainf/mirror1a.csv', delimiter = ',')
mirror1b = np.loadtxt('datainf/mirror1b.csv', delimiter = ',')
mirror2a = np.loadtxt('datainf/mirror2a.csv', delimiter = ',')
mirror2b = np.loadtxt('datainf/mirror2b.csv', delimiter = ',')

target_text = "flat" if flat else "cylindrical"
inverted_text = "inverted" if inv else "not inverted"
print(f"infinite system, {target_text} target, {inverted_text}, L = {L}, f = {f}, dB = {dB :.6f}, B_max = {B_max}")

plt.plot(mirror1a[:, 0], mirror1a[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1a')
plt.plot(mirror1b[:, 0], mirror1b[:, 1], linewidth = 1, c = 'cornflowerblue', label = 'mirror 1b')
plt.plot(mirror2a[:, 0], mirror2a[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2a')
plt.plot(mirror2b[:, 0], mirror2b[:, 1], linewidth = 1, c = 'mediumpurple', label = 'mirror 2b')

plt.axis('equal')
plt.legend()
plt.show()
```

```python
extr1 = np.loadtxt('datainf/extrinf1.csv', delimiter = ',')
extr2 = np.loadtxt('datainf/extrinf2.csv', delimiter = ',')

plt.plot([0, 0], [-0.005, 0.005], c = 'orange')

plt.plot(mirror1a[:, 0], mirror1a[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1a')
plt.plot(mirror1b[:, 0], mirror1b[:, 1], linewidth = 1, c = 'cornflowerblue', label = 'mirror 1b')
plt.plot(mirror2a[:, 0], mirror2a[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2a')
plt.plot(mirror2b[:, 0], mirror2b[:, 1], linewidth = 1, c = 'mediumpurple', label = 'mirror 2b')

for row in extr1: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'blue')
for row in extr2: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'red')

# plt.axis('equal')
plt.axis('square')
# zoom = 0.01
# plt.axis([-zoom, zoom, -zoom, zoom])
zoom = 0.0005
plt.axis([-zoom, zoom, 0.005 - zoom, 0.005 + zoom])
plt.show()
```

# Two-Mirror Concentrator

```python
# import
# data_2mc = np.loadtxt('data_2mc_flat.csv', delimiter = ',')
# data_2mc = np.loadtxt('data_2mc_cyl.csv', delimiter = ',')
data_2mc = np.loadtxt('data_2mc_fin_cyl.csv', delimiter = ',')

# parameters
f1, L, f2 = 2, 10, 1

# source and target
plt.scatter(-L, 0, s = 5, c = 'orange', label = 'source')
plt.scatter(0, 0, s = 5, c = 'red', label = 'target')

# M1
plt.plot(data_2mc[:, 0], data_2mc[:, 1], linewidth = 1, c = 'purple', label = 'M1')
# plt.text(data_2mc[0, 0], data_2mc[0, 1], s = f'({data_2mc[0, 0]}, {data_2mc[0, 1]})')
# plt.text(data_2mc[data_2mc.shape[0] - 1, 0], data_2mc[data_2mc.shape[0] - 1, 1], s = f'({data_2mc[data_2mc.shape[0] - 1, 0]}, {data_2mc[data_2mc.shape[0] - 1, 1]})')

# M2
plt.plot(data_2mc[:, 2], data_2mc[:, 3], linewidth = 1, c = 'blue', label = 'M2')
# plt.text(data_2mc[0, 2], data_2mc[0, 3], s = f'({data_2mc[0, 2]}, {data_2mc[0, 3]})')
# plt.text(data_2mc[data_2mc.shape[0] - 1, 2], data_2mc[data_2mc.shape[0] - 1, 3], s = f'({data_2mc[data_2mc.shape[0] - 1, 2]}, {data_2mc[data_2mc.shape[0] - 1, 3]})')

# parabola
M1Y = np.linspace(data_2mc[0, 1], data_2mc[-1, 1], 50)
M1X = 1 / (4 * f1) * (M1Y ** 2) - L - f1
plt.scatter(M1X, M1Y, s = 5, c = 'purple', label = 'PM1')
M2Y = np.linspace(data_2mc[0, 3], data_2mc[-1, 3], 50)
M2X = -1 / (4 * f2) * (M2Y ** 2) + f2
plt.scatter(M2X, M2Y, s = 5, c = 'blue', label = 'PM2')

# display
plt.axis('equal')
plt.legend()
plt.show()
```

# Concentration of mean rays

```python
# import
# data_mean_paths = np.loadtxt('data_2mc_flat_mean.csv', delimiter = ',')
data_mean_paths = np.loadtxt('data_2mc_cyl_mean.csv', delimiter = ',')

# zoom
plt.axis('square')
zoom = 0.001
plt.axis([-zoom, zoom, -zoom, zoom])

# circle
circle = plt.Circle((0, 0), zoom, color = 'black')
plt.gca().add_patch(circle)

# mean ray paths
for row in data_mean_paths:
    # X = [x for j, x in enumerate(row) if j % 2 == 0]
	# Y = [y for j, y in enumerate(row) if j % 2 == 1]
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'g', linewidth = 0.25)

# display
plt.show()
```

# Concentration of extreme rays

```python
# import
# data_pos_extr_paths = np.loadtxt('data_2mc_flat_extr_pos.csv', delimiter = ',')
# data_neg_extr_paths = np.loadtxt('data_2mc_flat_extr_neg.csv', delimiter = ',')
data_pos_extr_paths = np.loadtxt('data_2mc_cyl_extr_pos.csv', delimiter = ',')
data_neg_extr_paths = np.loadtxt('data_2mc_cyl_extr_neg.csv', delimiter = ',')

# zoom
plt.axis('square')
zoom = 0.01
plt.axis([-zoom, zoom, -zoom, zoom])

# circle
circle = plt.Circle((0, 0), zoom / 2.1, color = 'black')
plt.gca().add_patch(circle)

# extreme ray paths
for row in data_pos_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'r', linewidth = 0.25)
for row in data_neg_extr_paths:
    X = [row[4], row[6]]
    Y = [row[5], row[7]]
    plt.plot(X, Y, 'b', linewidth = 0.25)

# display
plt.show()
```
