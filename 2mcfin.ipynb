```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('figure', figsize = (8, 8))
```

# Two-Mirror Concentrator

```python
flat_source, flat_target, r, f1, L, f2, da, a_max = np.loadtxt('data_2mc_fin_input.csv', delimiter = ',')
data_2mc_m1a = np.loadtxt('data_2mc_fin_m1a.csv', delimiter = ',')
data_2mc_m1b = np.loadtxt('data_2mc_fin_m1b.csv', delimiter = ',')
data_2mc_m2a = np.loadtxt('data_2mc_fin_m2a.csv', delimiter = ',')
data_2mc_m2b = np.loadtxt('data_2mc_fin_m2b.csv', delimiter = ',')
# data_paths = np.loadtxt('data_2mc_fin_paths.csv', delimiter = ',')

source = 'flat' if flat_source else 'cylindrical'
target = 'flat' if flat_target else 'cylindrical'

if (f1 == f2): print(f'symmetric system: {source} source -> {target} target')
else: print(f'asymmetric system: {source} source -> {target} target')
print(f'radius = {r}, f1 = {f1}, L = {L}, f2 = {f2}, da = {da:.6f}, a_max = {a_max}')
```

```python
# source and target
if flat_source: plt.plot([-L, -L], [-r, r], linewidth = 5, color = 'orange', label = 'source')
else: plt.gca().add_patch(plt.Circle((-L, 0), r, color = 'orange', label = 'source'))
if flat_target: plt.plot([0, 0], [-r, r], linewidth = 5, color = 'red', label = 'target')
else: plt.gca().add_patch(plt.Circle((0, 0), r, color = 'red', label = 'target'))

# mirrors
plt.plot(data_2mc_m1a[:, 0], data_2mc_m1a[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1a')
plt.plot(data_2mc_m1b[:, 0], data_2mc_m1b[:, 1], linewidth = 1, c = 'cornflowerblue', label = 'mirror 1b')
plt.plot(data_2mc_m2a[:, 0], data_2mc_m2a[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2a')
plt.plot(data_2mc_m2b[:, 0], data_2mc_m2b[:, 1], linewidth = 1, c = 'mediumpurple', label = 'mirror 2b')

# parabolas
M1Y = np.linspace(data_2mc_m1a[-1, 1], data_2mc_m1b[-1, 1], 50)
M1X = 1 / (4 * f1) * (M1Y ** 2) - L - f1
plt.scatter(M1X, M1Y, s = 5, c = 'blue', label = 'parabola 1')
M2Y = np.linspace(data_2mc_m2a[-1, 1], data_2mc_m2b[-1, 1], 50)
M2X = -1 / (4 * f2) * (M2Y ** 2) + f2
plt.scatter(M2X, M2Y, s = 5, c = 'purple', label = 'parabola 2')

# paths
# for row in data_paths: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'green')

# display
plt.axis('equal')
plt.legend()
plt.show()
```

# Extreme Rays

```python
data_extr_1 = np.loadtxt('data_2mc_fin_extr_1.csv', delimiter = ',')
data_extr_2 = np.loadtxt('data_2mc_fin_extr_2.csv', delimiter = ',')
```

```python
if flat_target: plt.plot([0, 0], [-r, r], linewidth = 5, color = 'orange', label = 'target')
else: plt.gca().add_patch(plt.Circle((0, 0), 0.95 * r, color = 'orange', label = 'target'))

for row in data_extr_1: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.1, c = 'red')
for row in data_extr_2: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.1, c = 'blue')

plt.axis('square')
zoom = 0.25
plt.axis([-zoom, zoom, r - zoom, r + zoom])
# plt.axis([-zoom, zoom, -zoom, zoom])
plt.show()
```
