```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rc('figure', figsize = (10, 5))
```

# Two-Mirror Concentrator

```python
# import
f1, L, f2, da, a_max = np.loadtxt('data_2mc_fin_cyl_input.csv', delimiter = ',')
data_2mc_m1 = np.loadtxt('data_2mc_fin_cyl_m1.csv', delimiter = ',')
data_2mc_m2 = np.loadtxt('data_2mc_fin_cyl_m2.csv', delimiter = ',')
data_paths = np.loadtxt('data_2mc_fin_cyl_paths.csv', delimiter = ',')

# source and target
plt.scatter(-L, 0, s = 5, c = 'orange', label = 'source')
plt.scatter(0, 0, s = 5, c = 'red', label = 'target')

# mirrors
plt.plot(data_2mc_m1[:, 0], data_2mc_m1[:, 1], linewidth = 1, c = 'blue', label = 'mirror 1')
plt.plot(data_2mc_m2[:, 0], data_2mc_m2[:, 1], linewidth = 1, c = 'purple', label = 'mirror 2')

# parabolas
M1Y = np.linspace(data_2mc_m1[0, 1], data_2mc_m1[-1, 1], 50)
M1X = 1 / (4 * f1) * (M1Y ** 2) - L - f1
plt.scatter(M1X, M1Y, s = 5, c = 'blue', label = 'parabola 1')
M2Y = np.linspace(data_2mc_m2[0, 1], data_2mc_m2[-1, 1], 50)
M2X = -1 / (4 * f2) * (M2Y ** 2) + f2
plt.scatter(M2X, M2Y, s = 5, c = 'purple', label = 'parabola 2')

# paths
for row in data_paths: plt.plot([row[0], row[2]], [row[1], row[3]], linewidth = 0.2, c = 'green')

# display
plt.axis('equal')
plt.legend()
plt.show()
```
